```python
import pandas as pd
import numpy as np

url = 'https://raw.githubusercontent.com/YBI-Foundation/Dataset/main/Movies%20Recommendation.csv'
df = pd.read_csv(url)

display(df.shape)
display(df.head(3))
```

```python
features = ['Movie_Genre', 'Movie_Keywords', 'Movie_Tagline', 'Movie_Cast', 'Movie_Director']

for feature in features:
    df[feature] = df[feature].fillna('')

df['combined_features'] = df['Movie_Genre'] + ' ' + df['Movie_Keywords'] + ' ' + df['Movie_Tagline'] + ' ' + df['Movie_Cast'] + ' ' + df['Movie_Director']

display(df['combined_features'].head())
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

vectorizer = TfidfVectorizer()
feature_vectors = vectorizer.fit_transform(df['combined_features'])

similarity = cosine_similarity(feature_vectors)

display(similarity.shape)
display(similarity)
```

```python
import difflib

movie_name = input('Enter your favorite movie name: ')

list_of_all_titles = df['Movie_Title'].tolist()

find_close_match = difflib.get_close_matches(movie_name, list_of_all_titles)

close_match = find_close_match[0]

index_of_the_movie = df[df.Movie_Title == close_match].index[0]

similarity_score = list(enumerate(similarity[index_of_the_movie]))

sorted_similar_movies = sorted(similarity_score, key=lambda x: x[1], reverse=True)

print('\nTop 5 Movie Recommendations for you: \n')
for i, movie in enumerate(sorted_similar_movies[1:6], 1):
    index = movie[0]
    title_from_index = df[df.index == index]['Movie_Title'].values[0]
    print(f"{i}. {title_from_index}")
```
